```python
from typing import Self


class Vec:

    def __init__(self, src=None) -> Self:
        if src is None:
            self.elements = ()
        else:
            self.elements = tuple(src)

    def scalar_mult(self, alpha):
        el = list(self.elements)

        for i in range(len(el)):
            el[i] = el[i] * alpha

        return Vec(el)

    def __repr__(self):
        return repr(self.elements)


if __name__ == "__main__":
    v1 = Vec([10, 20, 30, 40, 50])
    v2 = v1.scalar_mult(3)

    print(v2)
    print(v1)
```

```python
#without using list
from typing import Self


class Vec:

    def __init__(self, src=None) -> Self:
        if src is None:
            self.elements = ()
        else:
            self.elements = tuple(src)

    def scalar_mult(self, alpha):
        return Vec(tuple(x * alpha for x in self.elements))

    def __repr__(self):
        return repr(self.elements)


if __name__ == "__main__":
    v1 = Vec([10, 20, 30, 40, 50])
    v2 = v1.scalar_mult(3)

    print(v2)
    print(v1)
```
